Mount Drive, verify 14 video ZIPs + map inventory, and prepare the runtime before the pilot.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
from pathlib import Path
import hashlib, shutil, zipfile
DRIVE_ROOT = Path('/content/drive/MyDrive/Video AIC')
BUNDLE_ZIP = DRIVE_ROOT / 'AIC25-Colab/aic25-colab-bundle-v1.zip'
BUNDLE_ROOT = Path('/content/aic25-colab-bundle-v1')
WORK_ROOT = Path('/content/aic25-work')
assert BUNDLE_ZIP.is_file(), f'Missing {BUNDLE_ZIP}'
expected = BUNDLE_ZIP.with_name(BUNDLE_ZIP.name + '.sha256').read_text().split()[0]
actual = hashlib.sha256(BUNDLE_ZIP.read_bytes()).hexdigest()
assert actual == expected, 'Bundle SHA-256 mismatch'
shutil.rmtree(BUNDLE_ROOT, ignore_errors=True)
BUNDLE_ROOT.mkdir(parents=True)
zipfile.ZipFile(BUNDLE_ZIP).extractall(BUNDLE_ROOT)
```
